```python
import pandas as pd

# Load the dataset (make sure the filename matches what you uploaded)
df = pd.read_csv('bank_loan.csv')

print("Dataset loaded successfully!")
```

```python
# Inspect the first five rows
print("First five rows of the dataset:")
display(df.head())

# Get a detailed summary of data types and non-null counts
print("\nDataFrame Info:")
df.info()
```

```python
import numpy as np

def convert_ccavg(val):
    if isinstance(val, str) and '/' in val:
        num, denom = val.split('/')
        try:
            d = float(denom)
            if d == 0:
                return 0.0
            return float(num) / d
        except (ValueError, ZeroDivisionError):
            return np.nan
    try:
        return float(val)
    except:
        return np.nan

# Apply the fixed conversion function
df['CCAvg'] = df['CCAvg'].apply(convert_ccavg)

# Verify the data type and sample values after cleaning
print("Updated CCAvg data type:", df['CCAvg'].dtype)
print("Sample cleaned CCAvg values:")
print(df['CCAvg'].head(10))

# Save the cleaned dataset
df.to_csv('bank_loan_cleaned.csv', index=False)
print("\nSuccess! Cleaned dataset saved as 'bank_loan_cleaned.csv'.")
```
