```python
# Install required libraries if not already available
!pip install ftfy contractions

import re
import string
import unicodedata
import ftfy
import contractions

# 1. Read the messy text file
file_path = 'text_sample.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    raw_text = file.read()

def clean_pharmaceutical_text(text):
    # Fix encoding issues using ftfy and unicodedata
    fixed_text = ftfy.fix_text(text)
    normalized_text = unicodedata.normalize('NFKD', fixed_text).encode('ascii', 'ignore').decode('utf-8')

    # Remove HTML tags if any exist
    html_free_text = re.sub(r'<.*?>', '', normalized_text)

    # Expand contractions (e.g., "don't" -> "do not")
    expanded_text = contractions.fix(html_free_text)

    # Remove random symbols/noise like @@, ## while keeping meaningful text
    symbol_cleaned = re.sub(r'[@#\$\%\^\&\*\_]+', ' ', expanded_text)

    # Convert text to lowercase for consistency
    lowercased = symbol_cleaned.lower()

    # Remove unnecessary punctuation (using string.punctuation)
    punctuation_free = lowercased.translate(str.maketrans('', '', string.punctuation))

    # Standardize specific variants or terms if found
    standardized = punctuation_free.replace("usd", "dollars").replace("$", "dollars")

    # Remove extra spaces, tabs, and newlines
    final_clean_text = re.sub(r'\s+', ' ', standardized).strip()

    return final_clean_text

# Apply the cleaning function
cleaned_text = clean_pharmaceutical_text(raw_text)

# Preview the results
print("--- Original Text Sample (First 300 chars) ---")
print(raw_text[:300])
print("\n--- Cleaned Text Sample (First 300 chars) ---")
print(cleaned_text[:300])

# 3. Save the cleaned text to a new file called cleaned_text.txt
output_path = 'cleaned_text.txt'
with open(output_path, 'w', encoding='utf-8') as output_file:
    output_file.write(cleaned_text)

print(f"\nSuccess! Cleaned text saved as '{output_path}'.")
```
