```python
import json
import pandas as pd

# Task 1: Load the JSON file into Python using the correct filename
file_path = 'sample.json'

with open(file_path, 'r') as file:
    data = json.load(file)

print("--- Task 1: JSON File Loaded Successfully ---")


# Task 2: Retrieve all order IDs along with their corresponding total amounts
print("\n--- Task 2: Order IDs and Total Amounts ---")
for record in data:
    if 'orders' in record:
        for order in record['orders']:
            print(f"Order ID: {order.get('order_id')} | Total Amount: {order.get('total')}")


# Task 3: Normalize the JSON structure using Pandas to create a flat DataFrame
# Based on sample.json, we unpack 'orders' while keeping 'customer_id'
df_flat = pd.json_normalize(
    data,
    record_path=['orders'],
    meta=['customer_id', 'name', 'email']
)

print("\n--- Task 3: Flattened DataFrame Preview ---")
print(df_flat.head())


# Task 4: Save the processed DataFrame back into a new JSON file
output_filename = 'processed_orders_output.json'
df_flat.to_json(output_filename, orient='records', indent=4)

print(f"\n--- Task 4: Success! Processed DataFrame saved as '{output_filename}' ---")
```
